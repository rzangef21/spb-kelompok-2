```python
import pandas as pd

df = pd.read_csv('/content/Dataset_sales.csv')
df.head()
```

```python
df['Unit_Price'] = pd.to_numeric(df['Unit_Price'], errors='coerce')
```

```python
df['Units_Sold'] = df['Units_Sold'].fillna(df['Units_Sold'].median())
df['Unit_Price'] = df['Unit_Price'].fillna(df['Unit_Price'].median())
```

```python
df.loc[df['Total_Revenue'].isna(), 'Total_Revenue'] = \
    df['Units_Sold'] * df['Unit_Price']
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
df['Transaction_ID'].duplicated().sum()
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
df['Sales_Level'] = pd.qcut(df['Total_Revenue'], q=3, labels=['Low','Medium','High'])
```

```python
monthly = df.groupby(pd.Grouper(key='Date', freq='ME'))['Total_Revenue'].sum().reset_index()
```

```python
monthly['Growth'] = monthly['Total_Revenue'].pct_change()

monthly['Status'] = monthly['Growth'].apply(
    lambda x: 'Naik' if x > 0 else ('Turun' if x < 0 else 'Stagnan')
)
```
